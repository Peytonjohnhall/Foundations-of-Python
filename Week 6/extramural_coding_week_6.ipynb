Create a class that can append strings from the StringIO module

```python
from io import StringIO

class StringBuilder:
    _file_str = None
    
    def __init__(self):
        self._file_str = StringIO()
        
    def Append(self, str):
        self._file_str.write(str)
        
    def __str__(self):
        return self._file_str.getvalue()
        
sb = StringBuilder()

sb.Append("Hello") 
sb.Append("World")

print(sb)
```

Using the format method

```python
# Example 1: named parameters
result1 = "{food} is the food of {user}".format(
	food="Ambrosia",
	user="the gods")
print(result1)

# Example 2: positional + indexing
result2 = "{0} is the food of {user[1]}".format(
	"Ambrosia",
	user=["men", "the gods", "others"]
)
print(result2)
```

STRING OPERATIONS Suppose that you have a list of strings in which some (but not necessarily all) of the strings begin and end with the double quote character:

x = ['"abc"', 'def', '"ghi"', '"klm"', 'nop']

What code would you use on each element to remove just the double quotes?

```python
x = ['"abc"', 'def', '"ghi"', '"klm"', 'nop']
x = [index.strip('"') for index in x]
print(x)
```

y = "Mississippi"

What code could you use to find the position of the last p in Mississippi? When you've found that position, what code would you use to remove just that letter?

```python
y = "Mississippi"

# find position of last 'p'
pos = y.rfind('p')
print(pos)

# remove that 'p'
y = y[:pos] + y[pos+1:]
print(y)
```

Look through these common string operations

```python
# base string
x = "hello"
y = "world"
# +
z = x + " " + y
print(z)
# *
z2 = " " * 20
print(z2)
# upper
print(x.upper())
# lower
print(x.lower())
# title
print("hello world".title())
# find, index
print(x.find("e"))
print(x.index("l"))
# rfind, rindex
s = "Mississippi"
print(s.rfind("p")) # far right p
print(s.rindex("p")) # far right p
print(s.rfind("i")) # far right i
print(s.rindex("i")) # far right i
# startswith, endswith
print(x.startswith("he"))
print(x.endswith("lo"))
# replace
print(x.replace("l", "X"))
# strip, rstrip, lstrip
s = "   hello   "
print(s.strip())
print(s.rstrip())
print(s.lstrip())
# encode
print(x.encode("utf_8"))
```
