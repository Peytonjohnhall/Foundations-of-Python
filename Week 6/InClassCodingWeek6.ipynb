```python
# importing copy module
import copy
# initializing list 1
li1 = [1, 2, [3,5], 4]
# using copy for shallow copy
li2 = copy.copy(li1)
print(id(li1))
print(li1)
print(id(li2))
print(li2)
li1[2].append(999)
print(id(li1))
print(li1)
print(id(li2))
print(li2)
```

```python
# importing copy module
import copy
# initializing list 1
li1 = [1, 2, [3,5], 4]
# using deepcopy for deepcopy
li2 = copy.deepcopy(li1)
print(id(li1))
print(li1)
print(id(li2))
print(li2)
li1[2].append(999)
print(id(li1))
print(li1)
print(id(li2))
print(li2)
```
