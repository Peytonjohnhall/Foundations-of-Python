```python
def example_fun(x, y, **other):
    print(other)
    print("x: {0}, y: {1}, keys in 'other': {2}".format(x, 
                    y, list(other.keys())))
    other_total = 0
    for k in other.keys(): # **other param is a dictionary
        other_total = other_total + other[k]
    print("The total of values in 'other' is {0}".format(other_total))
example_fun(2, y="1", foo=3, bar=4)
```

```python
def f(n, list1, list2):
    print("\n", "During the function operation: ", sep = "")
    print(f"ID of \"list1\" {type(list1)} (i.e. y) before appending: {id(list1)}")
    list1.append(3) # keeps id the same
    print(f"ID of \"list1\" {type(list1)} (i.e. y) after appending: {id(list1)}")
    print(f"ID of \"list2\" {type(list2)} (i.e. z) before defining the variable: {id(list2)}")
    list2 = [4, 5, 6] # creates a new object in memory
    print(f"ID of \"list2\" {type(list2)} (i.e. z) after defining the variable: {id(list2)}")
    print(f"ID of \"n\" {type(n)} (i.e. x) before adding 1: {id(n)}")
    n = n + 1 # creates a new object in memory
    print(f"ID of \"n\" {type(n)} (i.e. x) after adding 1: {id(n)}")
    
print("Before the function operation: ")
x = 5
print(f"Original ID of x {type(x)}: {id(x)}")
y = [1, 2]
print(f"Original ID of y {type(y)}: {id(y)}")
z = [4, 5]
print(f"Original ID of z {type(z)}: {id(z)}")

f(x, y, z) # call the function

print("\n", f"After the function operated: ", "\n",
      f"x {type(x)} = {x} at ID {id(x)}", "\n", 
      f"y {type(y)} = {y} at ID {id(y)}", "\n",
      f"z {type(z)} = {z} at ID {id(z)}", "\n", sep = "")
```

```python
def f_to_kelvin (degrees_f):
    return 273.15 + (degrees_f - 32) * 5 / 9
def c_to_kelvin(degrees_c):
    return 273.15 + degrees_c
abs_temperature = f_to_kelvin
print(abs_temperature(32))
abs_temperature = c_to_kelvin
print(abs_temperature(0))
t = {'FtoK': f_to_kelvin, 'CtoK': c_to_kelvin}
print(t['FtoK'](32))
print(t['CtoK'](0))
```

```python
def foo():
    name = "geek" # Our local variable
    def bar():
        nonlocal name # Reference name in the upper scope
        name = 'GeekForGeeks' # Overwrite this variable
        print(name)
    
    # Calling inner function
    bar()
    # Printing local variable
    print(name)
foo()
```

```python
def maximum(*numbers):
    if len(numbers) == 0:
        return None
    else:
        maxnum = numbers[0]
        for n in numbers[1:]:
            if n > maxnum:
                maxnum = n
    return maxnum

print(maximum(3, 2, 8))
print(maximum(1, 5, 9, -2, 2))
```
