Data Interning: Between -5 to 256, we have unique instances of numbers

```python
x = -5
y = -5
z = -5
print(id(x), id(y), id(z)) # all same
a = -6
b = -6
c = -6
print(id(a), id(b), id(c)) # all different
```

Between -5 to 256, it is called "intern".

```python
lst = [1, 2, 3]
print(id(lst))
lst.append(4)
print(id(lst))
lst = [1, 2, 3, 4]
print(id(lst))
lst = [1, 3]
print(id(lst))
```

Pass immutable types as parameters:

```python
def newFunction (aParm, bParm):
    # aParm, bParm are local variables not tweaked by reference
    aParm = aParm + 1
    bParm = bParm + 1
    return aParm + bParm
a = 10
b = 20
c = newFunction(a, b)
print(a)
print(b)
print(c)
```

Pass mutable types as parameters:

```python
def doSomethingtoRef(aRef):
    # aRef is being tweaked by reference (yet necessarily global)
    aRef.append(99)
lst = [1, 2, 3, 4, 5]
print(id(lst))
print(lst)
doSomethingtoRef(lst)
print(id(lst))
print(lst)
```

```python
def doFoo(x, y = 30):
    print(x + y)
doFoo(10, 20) # pass 10 for x and 20 for y
doFoo(10) # pass in 10 for x
def doFoo2(x = 100, y = 30):
    print(x+y)
doFoo2(10, 20) # pass 10 for x and 20 for y
doFoo2(10) # pass in 10 for x
doFoo2()
```

Positional argmemnts: one star. Name arguments: two stars.

```python
x = ' something'
def myfunc():
    global x
    x = "fantastic"
myfunc()
print("Python is " + x)
```

```python
x = ' something'
def myfuncc(x):
    x = "fantastic"
    return x
x = myfuncc(x)
print("Python is " + x)
```
