```python
def floats():
    bm = 8.3 + 3.5 * 4.4 / 10.6 # basic math
    fd = 6.5 // 3 # floor division
    mo = 20 % 2.5 # modulus
    eo = 2.2 ** 5 # exponentation
    return bm, fd, mo, eo

def main():
    bm, fd, mo, eo = floats()
    print("Basic math result", bm, "\n",
          "Floor division result = ", fd, "\n", 
          "Modulus result = ", mo, "\n", 
          "Exponentation result = ", eo)

if __name__ == "__main__":
    main()
```

```python
i = 1

while i < 6:
    print(i)
    i += 1 # same as i = i + 1
```

```python
# Exit the loop when i is 3:
i = 1

while i < 6:
    print(i)
    if i == 3:
        break   
    i += 1
```

```python
# Continue to the next iteration if i is 3:
i = 0

while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
```
