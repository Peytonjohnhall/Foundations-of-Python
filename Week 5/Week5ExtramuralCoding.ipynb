Modifying strings (https://www.w3schools.com/python/python_regex.asp)

```python
import re

txt = "The rain in Spain"
x = re.search("^The.*Spain$", txt) # search() Returns a Match object if there is a match anywhere in the string
print(x)
y = re.findall("ai", txt) # The findall() function returns a list containing all matches.
print(y)
z = re.findall("Portugal", txt)
print(z)
a = re.search("\s", txt) # \s refers to whitespace
print("The first white-space character is located in position:", a.start())
b = re.search("Portugal", txt)
print(b)
# The split() function returns a list where the string has been split at each match
c = re.split("\s", txt) # \s refers to whitespace
print(c)
d = re.split("\s", txt, 1) # only split once
print(d)
e = re.sub("\s", "9", txt)
print(e)
f = re.sub("\s", "9", txt, 2)
print(f)
g = re.search("ai", txt)
print(g) #this will print an object
# Print the string passed into the function:
h = re.search(r"\bS\w+", txt)
print(h.span())
# Print the part of the string where there was a match.
i = re.search(r"\bS\w+", txt)
print(i.group())
j = re.search(r"\bT\w+", txt)
print(j.group())
```

```python
txt = "Hello Sam!"
"""
The maketrans() method returns a mapping table that can be used with 
the translate() method to replace specified characters.
"""
mytable = str.maketrans("S", "P")
print(txt.translate(mytable))

txt = "Hi Sam!"
x = "mSa"
y = "eJo"
mytable = str.maketrans(x, y)
print(txt.translate(mytable))

txt = "Good night Sam!"
x = "mSa"
y = "eJo"
z = "odnght"
mytable = str.maketrans(x, y, z)
print(txt.translate(mytable))

txt = "I could eat bananas all day"
"""
The partition() method searches for a specified string, and splits 
the string into a tuple containing three elements.
"""
x = txt.partition("bananas")
print(x)

s1 = 'Geeks'.removeprefix("Gee")
print(s1)
s = "HelloWorld"
res = s.removeprefix("Hello")
print(res)  # Output: World
s = "PythonProgramming"
res = s.removeprefix("python") # Case Sensitivity
print(res)  # Output: PythonProgramming
s1 = "geeks.txt"
s2 = s1.removesuffix(".txt")
print(s2)
s1 = "geeky.TXT"
s2 = s1.removesuffix(".txt") # Case Sensitivity
print(s2)
txt = "I like bananas"
x = txt.replace("bananas", "apples")
print(x)
# Replace all occurrence of the word "one":
txt = "one one was a race horse, two two was one too."
x = txt.replace("one", "three")
print(x)
# Replace the two first occurrence of the word "one":
txt = "one one was a race horse, two two was one too."
x = txt.replace("one", "three", 2)
print(x)
# Where in the text is the last occurrence of the string "casa"?:
txt = "Mi casa, su casa."
x = txt.rindex("casa")
print(x)
# Where in the text is the last occurrence of the letter "e"?:
txt = "Hello, welcome to my world."
x = txt.rindex("e")
print(x)
# Return a 20 characters long, right justified version of the word "banana":
txt = "banana"
x = txt.rjust(20)
print(x, "is my favorite fruit.")
# Using the letter "O" as the padding character:
txt = "banana"
x = txt.rjust(20, "O")
print(x)
"""
The rpartition() method searches for the last occurrence of a specified string, 
and splits the string into a tuple containing three elements.
"""
txt = "I could eat bananas all day, bananas are my favorite fruit"
x = txt.rpartition("bananas")
print(x)
txt = "apple, banana, cherry"
x = txt.rsplit(", ")
print(x)
```

String searching example

```python
import re

log = """
Mission: Artemis III
Altitude: 384400 km
Velocity: 10.8 km/s
Fuel Remaining: 72%
Next Burn in 120 seconds
Radiation Level: 3.2 mSv
Signal Delay: 1.28 seconds
"""

print("=== Basic String Searching ===")

# Simple substring search
if "Velocity" in log:
    print("Velocity data found.")

# Find exact position
index = log.find("Fuel")
print("Position of 'Fuel':", index)

# Count occurrences of the word "seconds"
print("Occurrences of 'seconds':", log.count("seconds"))

print("\n=== Regular Expression Searching (re module) ===")

# Find all numbers
numbers = re.findall(r"\d+\.?\d*", log)
print("All numbers:", numbers)

# Find all measurements with units
measurements = re.findall(r"\d+\.?\d*\s?(km|km/s|mSv|seconds|%)", log)
print("Units found:", measurements)

# Extract altitude specifically
altitude = re.search(r"Altitude:\s*(\d+)\s*km", log)
if altitude:
    print("Altitude extracted:", altitude.group(1), "km")

# Find decimal values only
decimals = re.findall(r"\d+\.\d+", log)
print("Decimal values:", decimals)
```
