```python
# defining a decorator
def newDecoratorFunction(func):

	# inner1 is a Wrapper function in which the argument is called

	# inner function can access the outer local functions like in this case "func"
	def inner1():
		print(" 1 newDecoratorFunction has been invoked to decorate SomethingAlreadyExists() ")

		# calling the actual function now inside the wrapper function.
		func()

	print(" 3 Now I am decoratinhg the existing method SomethingAlreadyExists()")

	return inner1

# defining a function, to be called inside wrapper
def SomethingAlreadyExists():
	print(" 2 This is an existing function !! and is coming from SomethingAlreadyExists")

# passing 'function_to_be_used' inside the decorator to control its behavior
newDecoratedfunction = newDecoratorFunction(SomethingAlreadyExists)


# calling the function
newDecoratedfunction()
```

```python
def four():
	x = 0
	while x < 4:
		print("in generator, x =", x)
		yield x
		x += 1

for i in four():
	print(i, "hi")
```

```python
def make_pretty(func):
	def inner():
		print("I got decorated")
		func()
	return inner

def ordinary():
	print("I am ordinary")

# let's decorate this ordinary function
pretty = make_pretty(ordinary)
pretty()
```
