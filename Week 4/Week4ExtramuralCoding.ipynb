Play with immutable types:

```python
tu_ple = ({1, 2, 3}, "data")
tu_ple[0].add(4) # modify the set inside the tuple
print(tu_ple)
```

```python
t_uple = (["cat", "dog", "mouse"], 9, 22.5, True, {4, 5, 6}, "Green", {5: True, 6: False, 3: False})
t_uple[0].append("deer")
# t_uple[1].pop() # test. returns an error
# t_uple[2].remove() # test. returns an error
t_uple[4].add(7)
t_uple[6][12] = False
t_uple[6]["Eight"] = False
print(t_uple)
del t_uple[6]["Eight"]
print("\n", t_uple)
```

Play with sets:

```python
x = set([1, 2, 3, 1, 3, 5, 22, "q"])
print(x, "\n")
x.add(6)
print(x, "\n")
x.remove(5)
print(x, "\n")
print(1 in x, 4 in x, "\n")
y = set([1, 4, 5, 7, 8, 9, 22, "a"])
print(x | y) # "|" indicates Union
print(x & y) # "&" indicates Intersection
print(x ^ y) # "^" indicates symmetric difference
             # -that is, elements that are in one 
             # set or the other but not both.
```

Packing and unpacking tuples

```python
[a, b] = 11,21
print(a)
print(b)
[c, d] = 3, 4
print(c, d)
[e, f] = (5, 6)
print(e)
print(f)
(g, h) = 7, 8
print(g)
print(h)
i, j = [9, 10]
print(i)
print(j)
k, l = (11, 12)
print(k)
print(l)
print(a)
print(b, c, d)
print([b, c, d])
print((e, f, g))
print(h, i, j, k, l)
```
