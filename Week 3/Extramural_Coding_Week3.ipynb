```python
list1 = ["a", "b", ["aa", "bb", "cc", ["aaa", "bbb", "ccc"]]]
for eachItem in list1:
    for eachInnerItem in eachItem: # nested loop
        print(eachInnerItem)
        for eachInnerInnerItem in eachInnerItem: # nested loop inside a nested loop
            print(eachInnerInnerItem)
```

```python
l = [[22,33,44,55]] 
countItems = 0
for eachItem in l:
    countItems = countItems + len(eachItem)
    print(eachItem)
print (countItems)
```
