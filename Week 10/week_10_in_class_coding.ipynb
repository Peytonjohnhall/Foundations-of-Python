The purpose of a pickle file is to save python objects by converting them into binary format that can be saved and reaccessed later.

```python
import pickle
import datetime

list_of_names = ["Ali Naqvi", "Jon Doe", "Jim Schaffer"]
list_of_addresses = ["1234 Ali's lane", "345 John's address", "No known address"]
temperature_of_today = 34.25
date_of_today = datetime.datetime.now()

def pickleDump(list_of_names, list_of_addresses,
               temperature_of_today, date_of_today):
    file = open("pickle.pik", "wb") # write binary
    pickle.dump(list_of_names, file)
    pickle.dump(list_of_addresses, file)
    pickle.dump(temperature_of_today, file)
    pickle.dump(date_of_today, file)
    file.close()
    return file

def pickleLoad(file):
    file = open("pickle.pik", "rb") # read binary
    a = pickle.load(file)
    b = pickle.load(file)
    c = pickle.load(file)
    d = pickle.load(file)
    file.close()
    return a, b, c, d

file = pickleDump(list_of_names, list_of_addresses,
           temperature_of_today, date_of_today)
a, b, c, d = pickleLoad(file)
print(a, "\n", b, "\n", c, "\n", d, sep = "")
```

```python
# read the file, read each line,
# break it open, meaning split it,
# store all the attributes in different name attributes,
# has 6 attributes
# wrap it all into a function def Get_Retail_Item_Data()
def Get_Retail_Item_Data():

    file_object = open("RetailStoreItemData.txt", "r")

    Retail_Items = []

    first_line = True

    for eachLine in file_object:

        if first_line == True:
            first_line = False
            continue

        eachLine = eachLine.strip()

        a = eachLine.split(",")

        Item_UPC = a[0]

        # last 5 fields are always numeric → safe to take from end
        Item_Max_Qty = int(a[-5])
        Item_Order_Threshold = int(a[-4])
        Item_Replenishment_Order_Qty = int(a[-3])
        Item_On_Hand = int(a[-2])
        Item_Unit_Price = float(a[-1])

        # description = everything in between
        Item_Description = ",".join(a[1:-5])

        Retail_Items.append([Item_UPC,
                             Item_Description,
                             Item_Max_Qty,
                             Item_Order_Threshold,
                             Item_Replenishment_Order_Qty,
                             Item_On_Hand,
                             Item_Unit_Price])

    file_object.close()

    return Retail_Items
```

```python
Retail_Items = Get_Retail_Item_Data()
print(type(Retail_Items))
print(len(Retail_Items))
```
