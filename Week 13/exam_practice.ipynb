```python
original = [[1, 2], [3, 4]]

print("Original's contents: ", original)

shallow = original[:]

print("Shallow's contents: ", shallow)

print("Original's ID: ", id(original))
print("Shallow's ID: ", id(shallow)) 

shallow[0][0] = 99

print("Original's contents after editing shallow: ", original) 
print("Shallow's contents after editing shallow: ", shallow) 

print("Original's ID after editing shallow: ", id(original))
print("Shallow's ID after editing shallow: ", id(shallow)) 
print("Original's ID at index 0: ", id(original[0]))
print("Shallow's ID at index 0: ", id(shallow[0])) 
```

```python
import copy

original = [[1, 2], [3, 4]]
print("Original's contents: ", original)
deep = copy.deepcopy(original)
print("Deep's contents: ", deep)
print("Original's ID: ", id(original))
print("Deep's ID: ", id(deep))
deep[0][0] = 99
print("Original's contents after editing deep: ", original)
print("Deep's contents after editing deep: ", deep)
print("Original's ID after editing deep: ", id(original))
print("Deep's ID after editing deep: ", id(deep))
print("Original's ID at index 0: ", id(original[0]))
print("Deep's ID at index 0: ", id(deep[0]))
```
