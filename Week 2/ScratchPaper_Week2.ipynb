```python
# score = [99, 88, 77, 66, 78, 49, 99, 87, 96, 22, 89, 79, 74, 92, 91]
score = [99, 88, 77, 66, 78, 49, 99, 87, 96, 22, 89, 79, 22, 74, 92, 91]

loop_iteration_range = len(score)
first_to_disregard_hitherto = score[0] # assume conditional proof
second_to_disregard_hitherto = score[0] # assume conditional proof

def first_to_disregard(loop_iteration_range, first_to_disregard_hitherto):
    """
    Find the first score to disregard
    """
    for next_percentage in range(loop_iteration_range):
        if score[next_percentage] < first_to_disregard_hitherto:
            first_to_disregard_hitherto = score[next_percentage]
    first_to_disregard = first_to_disregard_hitherto
    return first_to_disregard

def second_to_disregard(score, loop_iteration_range, first_to_disregard, second_to_disregard_hithitherto):
    """
    Find the second score to disregard
    """
    min_score_count = 0
    for i in score:
        if i == first_to_disregard:
            min_score_count = min_score_count + 1
        if min_score_count == 2:
            break   # IMPORTANT: do not return 2; just stop counting

    if min_score_count == 1:
        def no_duplicate_minimums(score, loop_iteration_range, first_to_disregard, second_to_disregard_hitherto):
            for next_percentage in range(loop_iteration_range):
                if first_to_disregard < score[next_percentage] < second_to_disregard_hitherto:
                    second_to_disregard_hitherto = score[next_percentage]
            second_to_disregard = second_to_disregard_hitherto
            return second_to_disregard
        second_to_disregard = no_duplicate_minimums(score, loop_iteration_range, first_to_disregard, second_to_disregard_hithitherto)

    elif min_score_count == 2:
        def pair_of_minimums(score, loop_iteration_range, first_to_disregard, second_to_disregard_hitherto):
            """
            find the index of the left duplicate float,
            then find duplicate value at a higher index
            """
            left_duplicate_float_index = score.index(first_to_disregard)

            # find the next occurrence of the same minimum after the leftmost one
            for next_percentage in range(loop_iteration_range):
                if next_percentage > left_duplicate_float_index:
                    if score[next_percentage] == first_to_disregard:
                        right_duplicate_float = score[next_percentage]
                        return right_duplicate_float

            # fallback (should not happen if min_score_count == 2)
            return first_to_disregard

        second_to_disregard = pair_of_minimums(score, loop_iteration_range, first_to_disregard, second_to_disregard_hithitherto)

    return second_to_disregard

first_to_disregard_value = first_to_disregard(loop_iteration_range, first_to_disregard_hitherto)
second_to_disregard_value = second_to_disregard(score, loop_iteration_range, first_to_disregard_value, second_to_disregard_hitherto)

print(first_to_disregard_value, second_to_disregard_value)
score.remove(first_to_disregard_value)
score.remove(second_to_disregard_value)
score
```
