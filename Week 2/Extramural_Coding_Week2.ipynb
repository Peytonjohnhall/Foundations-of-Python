Let's test the dir() function

```python
class Person:
  name = "John"
  age = 36
  country = "Norway"

print(dir(Person))
```

Using nested "if ... elif ... else" statements in python:

```python
print("You have been halted by casino security!", "\n")
age = int(input("How old are you? "))
if age < 0:
        print("Error.")
if age >= 18:
    if age > 92:
        print("Keep walking. He does not care about your ID. You look old.")
    else:
        print("Let him see your ID.")
if 0 <= age < 18:
    print("You are too young to gamble!")
```
