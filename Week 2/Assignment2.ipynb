```python
import statistics

def get_scores():
    scores = []
    print("Enter exam scores one at a time. Once you are done, enter \"-1\"")
    
    while True:
        percentage = int(input())
    
        if 0 <= percentage <= 100:
            scores.append(percentage) # append, viz. add
        elif percentage == -1:
            break
        else:
            print("Invalid input")
    
    return scores

def calculate_average(scores):
    average = statistics.mean(scores) # see import statement
    return average

scores = get_scores()
if not scores:
    print("The list is empty.")
else:
    average = calculate_average(scores)
    print("The average score is:", "\n", average)
```
