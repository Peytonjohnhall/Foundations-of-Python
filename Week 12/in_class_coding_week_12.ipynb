```python
class doFoo:
    i = 123 # class var
    def __init__(self): # self references an object
        self.i = 99999 # instance var
```

```python
doFoo.i # access class var directly
```

```python
anObject = doFoo() # create instance
anObject.i # access instance var
```

```python
print(doFoo.i) # prints class var
print (doFoo().i) # prints instance var
```

static means you do not need to create an object in Python, also Java.

```python
class Music:
    @staticmethod
    def play():
        print("Playing music")
```

```python
Music.play()
```

no object was created, and nothing was returned in the static method

Java supports single class inheritance only. Python supports multiple class inheritance.

```python
class shape:
    def __init__(self, xCord, yCord): # parent class
        self.x = xCord # instance var
        self.y = yCord # instance var
        print("Base class shape", str(self.x), str(self.y))
```

```python
myShape = shape(10, 200) # create shape object
```

```python
class circle(shape): # inherits from shape
    def __init__(self, radius, x, y):
        print("inside the circle class now calling the super",
              "meaning the base class __init__ method", sep = "")
        super().__init__(x, y) # reference parent
        print("setting the value of radius in the child class")
        self.radius = radius # child-only var
```

```python
myCircle = circle(2.6, 100, 200) # create circle object
```

```python
class Person:
    def wear(self):
        print("I wear clothes")

class Man(Person):
    def wear(self): # overwrite the method
        print("I like pants")

class Woman(Person):
    def wear(self): # overwrite the method
        print("I like skirts")

class Kid(Person):
    # does not overwrite the method
    
    def who(self):
        print("I am just a kid")
```

```python
ali = Person()
ali.wear()
```

```python
Jim = Man()
Jim.wear()
```

```python
nina = Woman()
nina.wear()
```

```python
bill = Kid()
bill.wear()
```

```python
#!/usr/bin/python
class Employee:
    'Common base class for all employees'
    empCount = 0 # class var
    def __init__(self, name, salary): # __init__ initializes variables
        self.name = name # instance var
        self.salary = salary # instance var
        Employee.empCount += 1 # incrementation

    def displayCount(self): # self references an object
        print("Total Employee %d" % Employee.empCount)
        
    def displayEmployee(self):
        print("Name : ", self.name, ", Salary: ", self.salary)
        print("Employee.__doc__:", Employee.__doc__) # __doc__ returns and documents reference string of class
        print("Employee.__name__:", Employee.__name__) # __name__ calls the class name
        print("Employee.__module__:",Employee.__module__) # __module__ calls the module name consisting of class
        print("Employee.__bases__:", Employee.__bases__) # __bases__ prints the tuple including all the superclasses/ parents
        # "object will say: what are my bases?" - Professor Naqvi
        print("Employee.__dict__:", Employee.__dict__) # __dict__ returns a dict variable of class name space

objEmp = Employee("Ali", 30000) # define values the instance variables point to
objEmp.displayEmployee() # call the last function
```
