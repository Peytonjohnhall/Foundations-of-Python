```python
marks = 80
if marks >= 90:
    grade = 'A'
elif marks >= 80:
    grade = 'B'
elif marks >= 70:
    grade = 'C'
elif marks >= 60:
    grade = 'D'
else:
    grade = 'F'
print(grade)
```

```python
"""
If marks greater than or equal to 90 grade is "A"
If marks are in between 80 and 89 (inclusive) then grade is "B"
If marks are in between 70 and 79 (inclusive) then grade is "C"
If marks are in between 60 and 69 (inclusive) then grade is "D"
If marks below 60 then grade is "F"
"""
marks = 80
if marks >= 90:
    grade = 'A'
if 80 <= marks < 90:
    grade = 'B'
if 70 <= marks < 80:
    grade = 'C' 
if 60 <= marks < 70:
    grade = 'D'
if marks < 60:
    grade = 'F'
print(grade)
```

If the loop terminates prematurely with break, the else clause won't be executed.

```python
x = 6
while x:
    print(x)
    x -= 1
    if x == 3: 
        break
else:
    print('Done!')
```
