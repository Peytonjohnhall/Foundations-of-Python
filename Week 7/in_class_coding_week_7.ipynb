```python
# infinite loop without break
colors = ['red', 'green', 'blue']
for x in colors:
    if x == 'red':
        colors.insert(0, 'orange')
        print(colors)
        break
```

It is recommended that you first make a copy. 
The slicing operator makes this especially convenient.

```python
colors = ['red', 'green', 'blue']
for x in colors[:]: # make a copy of colors
    if x == 'red':
        colors.insert(0, 'orange')
print(colors)
# Prints ['orange', 'red', 'green', 'blue']
"""
Peyton's note: because we made a copy of colors, what we are 
actually iterating through is the original list, not the updated 
list. The reason why this distinction is important is because 
the for loop then does not iterate over an updated list and 
get stuck in an infinite loop. This infinite loop can come about
by getting stuck on x being equal to red (as seen in the
code block above this one).
"""
```

```python
# Loop through two lists at once
name = ['Bob', 'Sam', 'Max']
age = [25, 35, 30]
d = {}
zipped = zip(name, age)
print(zipped)
print(type(zipped))
for x, y in zip(name, age):
    print(x, y)
    d[x] = y
print(d.items())
# Prints Bob 25
# Prints Sam 35
# Prints Max 30
```
